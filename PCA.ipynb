```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
```

```python
data = {
    "StudyHours": [2, 4, 6, 8, 10],
    "Attendance": [60, 70, 80, 90, 100],
    "PreviousMarks": [55, 65, 75, 85, 95],
    "AssignmentMarks": [50, 60, 70, 80, 90]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

scaler = MinMaxScaler()

normalized_data = scaler.fit_transform(df)

normalized_df = pd.DataFrame(
    normalized_data,
    columns=df.columns
)

print("\nNormalized Data:")
print(normalized_df)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
data={
    "StudyHours":[1,2,3,4],
    "PreviousMarks":[2,3,5,4]
}
df=pd.DataFrame(data)
print("Orifinal Dataset:")
print(df)
```

```python
scaler = StandardScaler()

standardized_data = scaler.fit_transform(df)

standardized_df = pd.DataFrame(
    standardized_data,
    columns=df.columns
)

print("\nStandardized Dataset:")
print(standardized_df.round(6))
```

```python
covariance_matrix = np.cov(
    standardized_data,
    rowvar = False
)

print("\nCovariance Matrix:")
print(covariance_matrix.round(6))
```

```python
covariance_matrix = np.cov(
    standardized_data,
    rowvar=False
)

print("\nCovariance Matrix:")
print(covariance_matrix.round(6))
```

```python
eigenvalues, eigenvectors = np.linalg.eig(
    covariance_matrix
)

# Remove unnecessary complex-number notation
eigenvalues = eigenvalues.real
eigenvectors = eigenvectors.real

print("\nEigenvalues:")
print(eigenvalues.round(6))

print("\nEigenvectors:")
print(eigenvectors.round(6))
```

```python
index = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[index]
eigenvectors = eigenvectors[:, index]

print("\nSorted Eigenvalues:")
print(eigenvalues.round(6))

print("\nSorted Eigenvectors:")
print(eigenvectors.round(6))
```

```python
explained_variance = (
    eigenvalues / eigenvalues.sum()
)

print("\nExplained Variance Ratio:")
print(explained_variance.round(6))
```

```python
cumulative_variance = np.cumsum(
    explained_variance
)

print("\nCumulative Explained Variance:")
print(cumulative_variance.round(6))
```

```python
principal_components = np.dot(
    standardized_data,
    eigenvectors
)

pca_df = pd.DataFrame(
    principal_components,
    columns=["PC1", "PC2"]
)

print("\nPrincipal Components:")
print(pca_df.round(6))
```

```python
reduced_data = principal_components[:, 0]

reduced_df = pd.DataFrame(
    reduced_data,
    columns=["PC1"]
)

print("\nReduced Dataset:")
print(reduced_df.round(6))
```
